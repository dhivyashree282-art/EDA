```python
df["Medical Condition"] = df["Medical Condition"].fillna("Unknown")
```

```python
df["Billing Amount"] = df["Billing Amount"].fillna(df["Billing Amount"].median())
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
df["Admission Month"] = df["Date of Admission"].dt.month
df["Admission Date"] = df["Date of Admission"].dt.date
```

```python
df["Gender"] = df["Gender"].str.strip().str.title()
df["Medical Condition"] = df["Medical Condition"].str.strip().str.title()
df["Admission Type"] = df["Admission Type"].str.strip().str.title()
```

```python
def categorize_urgency(admission_type):
    if admission_type == "Emergency":
        return "Emergency"
    elif admission_type == "Urgent":
        return "Emergency"
    else:
        return "Elective"
df["Urgent"] = df["Admission Type"].apply(categorize_urgency)
```

```python
print(df["Urgency"].value_counts())
```

```python
print(df["Gender"].value_counts())
```

```python
print(df["Gender"].value_counts(normalize=True) * 100)
```

```python
print(df["Age"].describe())
```

```python
bins = [0, 18, 35, 50, 65, 100]
labels = ["0-18", "19-35", "36-50", "51-65", "66+"]

df["Age Group"] = pd.cut(
    df["Age"],
    bins=bins,
    labels=labels
)
```

```python
df["Age Group"].value_counts()
```

```python
df["Billing Amount"].describe()
```

```python
print("Mean:", df["Billing Amount"].mean())
print("Median:", df["Billing Amount"].median())
print("Minimum:", df["Billing Amount"].min())
print("Maximum:", df["Billing Amount"].max())
print("Standard Deviation:", df["Billing Amount"].std())
```

```python
df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)
```

```python
df["Stay Days"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days
```

```python
df["Stay Days"].describe()
```

```python
df["Medical Condition"].value_counts()
```

```python
gender_condition = pd.crosstab(
    df["Medical Condition"],
    df["Gender"]
)

print(gender_condition)
```

```python
age_condition = pd.crosstab(
    df["Medical Condition"],
    df["Age Group"]
)

print(age_condition)
```

```python
stay_condition = df.groupby(
    "Medical Condition"
)["Stay Days"].mean()

print(stay_condition)
```
