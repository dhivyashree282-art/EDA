```python
print("Zero Bedrooms:", (df["bedrooms"] == 0).sum())
print("Zero Bathrooms:", (df["bathrooms"] == 0).sum())
```

```python
df["bedrooms"] = df["bedrooms"].replace(0, df["bedrooms"].median())

df["bathrooms"] = df["bathrooms"].replace(0, df["bathrooms"].median())

print("Zero Bedrooms:", (df["bedrooms"] == 0).sum())
print("Zero Bathrooms:", (df["bathrooms"] == 0).sum())
```

```python
df["date"] = pd.to_datetime(df["date"])

print(df["date"].dtype)
```

```python
df["Sale Year"] = df["date"].dt.year

df[["date", "Sale Year"]].head()
```

```python
df["Property Age"] = df["Sale Year"] - df["yr_built"]

df[["yr_built", "Sale Year", "Property Age"]].head()
```

```python
df["Is Renovated"] = df["yr_renovated"].apply(lambda x: "Yes" if x > 0 else "No")

df[["yr_renovated", "Is Renovated"]].head()
```

```python
df["Price per Sqft Living"] = df["price"] / df["sqft_living"]

df[["price", "sqft_living", "Price per Sqft Living"]].head()
```
