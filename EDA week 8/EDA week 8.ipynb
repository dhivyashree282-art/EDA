```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Cleaned_StudentsPerformance.csv")
```

```python
df.groupby("parental level of education")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("parental level of education")[["math score", "reading score", "writing score"]].mean().plot(kind="bar", figsize=(12,6))

plt.title("Average Scores Based on Parental Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Scores")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
df.groupby("lunch")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("lunch")[["math score", "reading score", "writing score"]].mean().plot(kind="bar", figsize=(8,5))

plt.title("Student Performance Based on Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Scores")
plt.show()
```

```python
df_melted = df.melt(id_vars="test preparation course", value_vars=["math score", "reading score", "writing score"], var_name="Subject", value_name="Score")

plt.figure(figsize=(10,6))
sns.boxplot(data=df_melted, x="test preparation course", y="Score", hue="Subject")
plt.title("Scores Based on Test Preparation Course")
plt.xlabel("Test Preparation Course")
plt.ylabel("Scores")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["math score"], df["reading score"])
plt.title("Math vs Reading Scores")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["math score"], df["writing score"])
plt.title("Math vs Writing Scores")
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["reading score"], df["writing score"])
plt.title("Reading vs Writing Scores")
plt.xlabel("Reading Score")
plt.ylabel("Writing Score")
plt.show()
```

```python
df[["math score", "reading score", "writing score"]].corr()
```

```python
plt.figure(figsize=(7,5))
sns.heatmap(df[["math score", "reading score", "writing score"]].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
correlation = df[["math score", "reading score", "writing score"]].corr()

correlation_values = correlation.where(np.triu(np.ones(correlation.shape), k=1).astype(bool))

strongest_correlation = correlation_values.stack().idxmax()

print("Strongest Correlation:", strongest_correlation)
print("Correlation Value:", correlation_values.stack().max())
```
